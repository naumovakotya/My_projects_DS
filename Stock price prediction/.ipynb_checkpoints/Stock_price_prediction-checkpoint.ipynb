```python
def download_and_merge_stocks(stock_dict, start_date, end_date):
    """Function for downloding data and merging all list of stocks"""
    # Создаем пустой список для хранения отдельных DataFrame
    merged_data = []

    for name, ticker in stock_dict.items():
        # Загружаем данные
        stock_data = yf.download(ticker, start=start_date, end=end_date, progress=False)
        
        # Переименовываем столбцы
        stock_data = stock_data.add_suffix(f"_{name}")
        stock_data = stock_data.rename(columns={f'Adj Close_{name}': f'Adj_Close_{name}'})
        
        # Добавляем полученные данные в список
        merged_data.append(stock_data)
    
    # Объединяем все данные по индексу (дата)
    merged_df = pd.concat(merged_data, axis=1)
    
    return merged_df
```

Краткие пояснения, что за собой скрывают названия столбцов:

1. **Date**: Дата торгового дня.

2. **Open**: Цена открытия акций в начале торгового дня.

3. **High**: Самая высокая цена, достигнутая акциями в течение торгового дня.

4. **Low**: Самая низкая цена, достигнутая акциями в течение торгового дня.

5. **Close**: Цена закрытия акций в конце торгового дня (без учета корпоративных действий).

6. **Adj Close**: Скорректированная цена закрытия, которая учитывает все корпоративные действия, такие как выплата дивидендов, разделение акций и т. д.

7. **Volume**: Объем торгов, показывающий количество акций, которые были куплены и проданы в течение торгового дня.

### Functions

```python
# Feature engenering
def new_features(df, name=NAME, sma_periods=[3, 5, 10], target=TARGET, max_lag=13):
    """
    Function is adding new features like:
    - differences between all basik columns, 
    - calculated SMA (Simple Moving Average), EMA (Exponential Moving Average) and Volatility (sma_periods)
    - Volume and Price change
    - Features of trend and seasonality
    - Lag features (max_lag)
    - Create Target (target), which is .shift(-1) of target_col
    """
    # Приведение всех столбцов к общему виду
    target_col = f'{target}_{name}'
    
    # Скользящие средние (SMA и EMA) и Волатильность
    for period in sma_periods:
        df[f'SMA_{period}_{name}'] = df[target_col].rolling(window=period).mean()
        df[f'EMA_{period}_{name}'] = df[target_col].ewm(span=period, adjust=False).mean()
        df[f'Volatility_{period}_{name}'] = df[target_col].rolling(window=period).std()
    
    # Процентные изменения
    df[f'Volume_Change_{name}'] = df[f'Volume_{name}'].pct_change()
    df[f'Price_Change_{name}'] = df[target_col].pct_change()
    
    # Создание базовых признаков (на базе разности)
    df[f'Open_High_{name}'] = df[f'Open_{name}'] - df[f'High_{name}']
    df[f'Open_Low_{name}'] = df[f'Open_{name}'] - df[f'Low_{name}']
    df[f'High_Low_{name}'] = df[f'High_{name}'] - df[f'Low_{name}']
    df[f'High_Close_{name}'] = df[f'High_{name}'] - df[f'Close_{name}']
    df[f'Low_Close_{name}'] = df[f'Low_{name}'] - df[f'Close_{name}']
    df[f'Close_Open_{name}'] = df[f'Close_{name}'] - df[f'Open_{name}']
    df[f'AdjClose_Open_{name}'] = df[f'Adj_Close_{name}'] - df[f'Open_{name}']
    df[f'AdjClose_Low_{name}'] = df[f'Adj_Close_{name}'] - df[f'Low_{name}']
    df[f'AdjClose_High_{name}'] = df[f'Adj_Close_{name}'] - df[f'High_{name}']
    df[f'AdjClose_Close_{name}'] = df[f'Adj_Close_{name}'] - df[f'Close_{name}']
    
    # Добавление признаков тренда и сезонности
    df[f'Day_of_Week_{name}'] = df.index.dayofweek
    df[f'Number_of_Week_{name}'] = df.index.isocalendar().week
    df[f'Month_{name}'] = df.index.month
    df[f'Quarter_{name}'] = df.index.quarter
    
    # Добавление лаг-функций
    for lag in range(1, max_lag):
        df[f'Lag_{lag}_{name}'] = df[target_col].shift(lag)
    
    # Добавляем столбец с ценой таргета на следующий день
    df['Target'] = df[target_col].shift(-1)
    df['Low_future'] = df[f'Low_{name}'].shift(-1)
    
    # Удаление строк с NaN значениями, которые появились из-за использования скользящих средних и других вычислений
    df.dropna(inplace=True)
    
    return df
```

```python
# Визуализация данных
class StockDataVisualizer:
    def __init__(self, data, label_suffix=None):
        """
        Инициализация класса.
        
        :param data: DataFrame с данными о ценах акций.
        :param label_suffix: Суффикс для добавления в заголовки графиков (например, "train" или "test").
        """
        self.data = data
        self.label_suffix = f"({label_suffix})" if label_suffix else ""
    
    def plot_time_series(self, columns, labels, title, ylabel, xlabel='Date', figsize=(12, 6)):
        """
        Построение графиков временных рядов.
        :param columns: Список столбцов для построения.
        :param labels: Список меток для легенды.
        :param title: Заголовок графика.
        :param ylabel: Подпись оси Y.
        :param xlabel: Подпись оси X.
        :param figsize: Размер графика.
        """
        plt.figure(figsize=figsize)
        for col, label in zip(columns, labels):
            plt.plot(self.data.index, self.data[col], label=label)
        plt.title(title)
        plt.xlabel(xlabel)
        plt.ylabel(ylabel)
        plt.legend()
        plt.show()

    def plot_correlation_heatmap(self, columns, title='Correlation Heatmap', figsize=(8, 6)):
        """
        Построение тепловой карты корреляции.
        :param columns: Список столбцов для анализа корреляции.
        :param title: Заголовок графика.
        :param figsize: Размер графика.
        """
        corr = self.data[columns].corr()
        plt.figure(figsize=figsize)
        sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
        plt.title(title)
        plt.show()

    def plot_grouped_bar(self, group_by_column, value_column, agg_func='mean', title='', xlabel='', ylabel='', figsize=(10, 6)):
        """
        Построение сгруппированных столбчатых диаграмм.
        :param group_by_column: Столбец для группировки.
        :param value_column: Столбец с числовыми данными.
        :param agg_func: Агрегационная функция ('mean', 'sum', и т.д.).
        :param title: Заголовок графика.
        :param xlabel: Подпись оси X.
        :param ylabel: Подпись оси Y.
        :param figsize: Размер графика.
        """
        grouped_data = self.data.groupby(group_by_column)[value_column].agg(agg_func)
        grouped_data.plot(kind='bar', figsize=figsize)
        plt.title(title)
        plt.xlabel(xlabel)
        plt.ylabel(ylabel)
        plt.show()

    def plot_with_secondary_axis(self, company, target, sma_period=10, ema_period=10, 
                                  secondary_metric=None, secondary_axis=False, data=None):
        """
        Построение графика с SMA, EMA и опциональной вторичной осью (волатильность или объем).
        
        :param company: Название компании (например, "apple").
        :param target: Целевая переменная (например, "Open").
        :param sma_period: Период для SMA.
        :param ema_period: Период для EMA.
        :param secondary_metric: Метрика для вторичной оси (например, "Volatility" или "Volume").
        :param secondary_axis: Включить вторичную ось (True/False).
        :param data: Используемый DataFrame (по умолчанию будет использоваться `self.data`).
        """
        data = data if data is not None else self.data

        plt.figure(figsize=(12, 6))

        # Основной график: Цены и SMA/EMA
        plt.plot(data[f'{target}_{company}'], label=f'{company.capitalize()} {target}', color='blue')
        plt.plot(data[f'SMA_{sma_period}_{company}'], label=f'SMA {sma_period}', color='orange')
        plt.plot(data[f'EMA_{ema_period}_{company}'], label=f'EMA {sma_period}', color='green')

        if secondary_axis and secondary_metric:
            # Вторичная ось
            secondary_column = f'{secondary_metric}_{sma_period}_{company}' \
                if secondary_metric == 'Volatility' else f'{secondary_metric}_{company}'
            
            ax2 = plt.gca().twinx()
            ax2.plot(data[secondary_column], label=secondary_metric, color='red', linestyle='--', alpha=0.7)
            ax2.set_ylabel(secondary_metric, color='red')
            ax2.tick_params(axis='y', labelcolor='red')

        # Настройки графика
        plt.title(f'{target} Prices with SMA, EMA for {company.capitalize()} {self.label_suffix}')
        plt.xlabel('Date')
        plt.ylabel('Price')
        plt.legend(loc='upper left')
        plt.grid(True)
        plt.show()
    
    def plot_last_years(self, company, target, sma_period=10, ema_period=10, 
                        secondary_metric=None, secondary_axis=False, years=1):
        """
        Построение графика для анализа данных за последние N лет с использованием метода `plot_with_secondary_axis`.

        :param company: Название компании (например, "apple").
        :param target: Целевая переменная (например, "Open").
        :param sma_period: Период для SMA.
        :param ema_period: Период для EMA.
        :param secondary_metric: Метрика для вторичной оси (например, "Volatility" или "Volume").
        :param secondary_axis: Включить вторичную ось (True/False).
        :param years: Количество последних лет для анализа.
        """
        # Определяем временной промежуток
        end_date = self.data.index.max()  # Последняя доступная дата
        start_date = end_date - pd.DateOffset(years=years)  # Начало анализа

        # Фильтруем данные по времени
        filtered_data = self.data[(self.data.index >= start_date) & (self.data.index <= end_date)]

        # Вызов метода класса с фильтрацией
        self.plot_with_secondary_axis(
            company=company,
            target=target,
            sma_period=sma_period,
            ema_period=ema_period,
            secondary_metric=secondary_metric,
            secondary_axis=secondary_axis,
            data=filtered_data
        )
```

```python
# Визуализация данных одновременно нескольких экземпляров класса
def process_multiple_instances(instances, method_name, **kwargs):
    """
    Обработка нескольких экземпляров класса `StockDataVisualizer`.

    :param instances: Список экземпляров класса `StockDataVisualizer`.
    :param method_name: Название метода для вызова.
    :param kwargs: Аргументы для метода.
    """
    for i, instance in enumerate(instances):
        method = getattr(instance, method_name)
        # Добавляем уточнение в заголовок графика в зависимости от типа данных
        data_type = "Training Data" if i == 0 else "Test Data"  # Первый экземпляр — тренировочные данные, второй — тестовые
        if 'title' in kwargs:
            kwargs['title'] = f"{kwargs['title']} ({data_type})"
        method(**kwargs)
```

### Feature engineering

На этапе **feature engineering** создаются дополнительные признаки, которые обогащают исходные данные для улучшения предсказательной способности модели. Добавленные признаки помогут выявить скрытые закономерности, связанные с трендами, волатильностью, сезонностью и историческими данными. 

Особое внимание уделяется тому, чтобы избежать утечек данных. Все создаваемые признаки основаны исключительно на исторической информации, доступной на момент предсказания. Например:
- Лаговые признаки (`Lag_1`, `Lag_2`, и т.д.) содержат информацию о предыдущих значениях целевой переменной.
- Признаки трендов (скользящие средние `SMA`, экспоненциальные средние `EMA`) и волатильность рассчитываются на основе доступных данных за предшествующие периоды.
- Переменная `df['Low_future']` добавлена для дополнительного анализа, но **не будет использоваться для обучения модели**, чтобы исключить риск утечки данных.

Эти меры гарантируют, что модель будет работать корректно и надежно при предсказании на тестовых данных и новых наблюдениях.

На этом этапе реализуем генерацию дополнительных признаков для обогащения данных с целью предсказания **цены открытия акций компании Apple на следующий день**. Ниже представлен подробный разбор добавленных признаков.

---

**1. Базовые признаки: разности между основными показателями**

Анализ внутридневной и междневной динамики цен. Эти признаки помогают модели учитывать волатильность и изменения цен в течение дня. Стоит также отменить, что данные признаки были созданы не только для целевой переменной, но и для других компаний.

**Признаки**:
- `Open_High`: Разница между ценой открытия (`Open`) и максимальной ценой (`High`).
- `Open_Low`: Разница между ценой открытия и минимальной ценой (`Low`).
- `High_Low`: Дневной диапазон цен (разница между `High` и `Low`).
- `High_Close`: Разница между максимальной ценой (`High`) и ценой закрытия (`Close`).
- `Low_Close`: Разница между минимальной ценой (`Low`) и ценой закрытия (`Close`).
- `Close_Open`: Разница между ценой закрытия (`Close`) и ценой открытия (`Open`).
- `AdjClose_Open`: Разница между скорректированной ценой закрытия (`Adj Close`) и ценой открытия (`Open`).
- `AdjClose_Low`: Разница между скорректированной ценой закрытия (`Adj Close`) и минимальной ценой (`Low`).
- `AdjClose_High`: Разница между скорректированной ценой закрытия (`Adj Close`) и максимальной ценой (`High`).
- `AdjClose_Close`: Разница между скорректированной ценой закрытия (`Adj Close`) и ценой закрытия (`Close`).

Базовые признаки формируют основу для анализа поведения акций и включают ключевые различия между основными метриками (`Open`, `High`, `Low`, `Close`, `Adj Close`). Например:
- **`Open_High`** позволяет оценить разницу между ценой открытия и максимальной ценой, что может указывать на восходящий тренд.
- **`AdjClose_Low`** помогает учитывать влияние корпоративных действий на минимальную цену в течение дня.

Эти признаки создают дополнительный контекст, который особенно важен для задач временных рядов, где даже небольшие изменения могут быть значимыми для предсказания.

---

**2. Скользящие средние (SMA, EMA) и волатильность**

Выявление трендов и сглаживание данных для получения более информативных признаков. Скользящие средние показывают краткосрочные тренды, а волатильность — уровень риска.

**Признаки**:
  - **SMA** (Simple Moving Average):
    - Примеры: `SMA_3`, `SMA_5`, `SMA_10` — среднее значение за последние 3, 5 и 10 дней.
  - **EMA** (Exponential Moving Average):
    - Похоже на SMA, но с большим весом для недавних данных.
  - **Волатильность** (Volatility):
    - Стандартное отклонение целевой переменной (`Open`) за последние 3, 5 или 10 дней.

---

**3. Процентные изменения**

Анализ относительных изменений объёма торговли и цены. Эти признаки важны для оценки динамики рынка и активности.

**Признаки**:
  - `Volume_Change`: Процентное изменение объёма торговли за день.
  - `Price_Change`: Процентное изменение цены открытия (`Open`) по сравнению с предыдущим днём.
  
---

**4. Признаки тренда и сезонности**

Учёт временных паттернов. Эти признаки позволяют модели учитывать сезонные тренды и их влияние на рынок.

**Признаки**:
  - `Day_of_Week`: День недели (например, понедельник = 0, пятница = 4).
  - `Number_of_Week`: Номер недели года.
  - `Month`: Номер месяца.
  - `Quarter`: Квартал.

---

**5. Лаг-функции**

Предоставление исторического контекста для предсказания. Лаг-признаки важны для анализа временной зависимости, что критично для временных рядов.

**Признаки**:
  - `Lag_1`, `Lag_2`, ..., `Lag_12`: Значения цены открытия (`Open`) за последние 12 дней.

---

**6. Целевая переменная**

Определить метрику, которую нужно предсказывать.

**Признаки**:
  - `Target`: Цена открытия на следующий день (`shift(-1)`).
  - `Low_future`: Минимальная цена на следующий день (`shift(-1)`).

---

**7. Удаление строк с пропусками**

Строки с `NaN`, возникающие из-за вычисления скользящих средних, лагов и других признаков, удаляются для обеспечения корректности данных.

```python
def optimize_df(df):
    """Уменьшение размерности датафрейма"""
    initial_memory = df.memory_usage(deep=True).sum()
    
    for col in df.columns:
        col_type = df[col].dtype
        
        if col_type == 'float64' or col_type == 'float32':
            # Проверяем, можно ли привести float к int
            if df[col].apply(float.is_integer).all():
                df[col] = pd.to_numeric(df[col], downcast='integer')
            else:
                df[col] = pd.to_numeric(df[col], downcast='float')
        
        elif col_type == 'int64' or col_type == 'int32':
            df[col] = pd.to_numeric(df[col], downcast='integer')
        
        elif col_type == 'object':
            num_unique_values = len(df[col].unique())
            num_total_values = len(df[col])
            if num_unique_values / num_total_values < 0.5:
                df[col] = df[col].astype('category')
        
        elif col_type == 'bool':
            df[col] = df[col].astype('int8')
    
    final_memory = df.memory_usage(deep=True).sum()
    memory_reduction = (initial_memory - final_memory) / initial_memory * 100
    
    print(f"Память до оптимизации: {initial_memory} байт")
    print(f"Память после оптимизации: {final_memory} байт")
    print(f"Сокращение памяти: {memory_reduction:.2f}%")
    
    return df
```

```python
def plot_feature_importances(model, feature_names, model_name):
    """
    Функция для вывода важности признаков для модели.

    Parameters:
    model (sklearn model): Обученная модель.
    feature_names (list): Список названий признаков.
    model_name (str): Название модели.
    """
    if hasattr(model, 'feature_importances_'):
        importances = model.feature_importances_
    elif hasattr(model, 'coef_'):
        importances = model.coef_
    else:
        raise ValueError(f"Model {model_name} does not have feature importances or coefficients.")

    # Создание DataFrame для удобного отображения
    feature_importance_df = pd.DataFrame({
        'Feature': feature_names,
        'Importance': importances
    })
    feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)
    
    # Построение графика
    plt.figure(figsize=(10, 6))
    sns.barplot(x='Importance', y='Feature', data=feature_importance_df)
    plt.title(f'Feature Importances for best model - {model_name}')
    plt.show()
```
